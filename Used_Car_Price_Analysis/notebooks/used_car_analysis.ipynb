```python
# Final Project Insights

## Key Findings

1. **Car Age and Selling Price**
   
   Newer cars generally have higher selling prices than older cars. The correlation between `year` and `selling_price` is approximately **0.414**, indicating a moderate positive relationship.

2. **Kilometers Driven and Selling Price**
   
   `km_driven` has a correlation of approximately **-0.226** with `selling_price`. This indicates a weak negative relationship, meaning cars with higher kilometers driven tend to have somewhat lower selling prices.

3. **Number of Seats and Selling Price**
   
   The correlation between `seats` and `selling_price` is approximately **0.041**, which is very close to zero. This suggests that the number of seats has little linear relationship with selling price.

4. **Seller Type and Average Selling Price**
   
   Dealer-listed cars have the highest average selling price at approximately **₹926,722**, followed by Trustmark Dealers at approximately **₹718,111**. Individual sellers have the lowest average selling price at approximately **₹471,555**.

5. **Car Features and Selling Price**
   
   The Linear Regression model shows positive coefficients for `mileage`, `engine`, and `max_power`, while `km_driven` and `car_age` have negative coefficients. This indicates that these features contribute differently to predicted selling prices.

6. **Effect of Car Age**
   
   The Linear Regression coefficient for `car_age` is approximately **-39,892**. This indicates that, while keeping other features constant, an increase of one year in car age is associated with a decrease of approximately **₹39,892** in the model's predicted selling price.

7. **Effect of Maximum Power**
   
   The coefficient for `max_power` is approximately **15,574**, showing a positive relationship with predicted selling price in the Linear Regression model.

8. **Machine Learning Model Performance**
   
   The Linear Regression model achieved an **R² score of 0.6559**, meaning that approximately **65.6% of the variation in car selling prices is explained by the features used in the model**.

9. **Prediction Error**
   
   The model achieved a **Mean Absolute Error (MAE) of approximately ₹275,718**. This means that the model's predicted selling prices differ from the actual selling prices by approximately ₹2.76 lakh on average.

10. **Model Limitations**
    
    The Actual vs Predicted plot shows that the model captures the general relationship between actual and predicted prices, but there is considerable variation in some predictions, particularly for higher-priced vehicles. The model also produced some negative predicted prices, which are unrealistic and demonstrate a limitation of using simple Linear Regression for this dataset.

## Machine Learning Results

- **Model:** Linear Regression
- **Features Used:** `km_driven`, `mileage`, `engine`, `max_power`, `car_age`, `seats`
- **Target:** `selling_price`
- **MAE:** ₹275,717.72
- **R² Score:** 0.6559
- **Explained Variation:** Approximately 65.6%

## Overall Conclusion

The analysis shows that **car age, kilometers driven, engine characteristics, and maximum power are associated with used car selling prices**. Newer cars generally command higher prices, while cars with higher kilometers driven tend to have lower prices. Seller type also shows noticeable differences in average selling price.

The Linear Regression model provides a reasonable baseline, explaining approximately **65.6% of the variation in selling prices**. However, incorporating additional variables such as **brand, car model, fuel type, transmission, and seller type** could potentially improve the model's predictive performance.
```
